# Variables y expresiones aritméticas


## Variables

Las variables son elementos presentes en todos los lenguajes de programación y nos permiten almacenar información para ser reutilizada cuando lo necesitemos a lo largo de nuestro código.

Para declarar una variable basta con escribir un **identificador o nombre y asignarle un valor usando el símbolo `=`**. 

Por ejemplo, podemos almacenar una dirección web:

```python
url_kaggle = "https://kaggle.com/"
```

y consultarla cuando lo necesitemos

```python
url_kaggle
```

En Python podemos almacenar **cualquier elemento** en una variable. En esta libreta vamos a trabajar con dos tipos básicos:

- Números
- Textos

Los números se representan introduciendo el valor numérico directamente, el cual puede contener o no parte decimal:

```python
125
```

```python
0.95
```

Los textos se representan introduciendo los caracteres encerrados entre los símbolos `"` o `'`:

```python
"Esto es un texto"
```

```python
'y esto también'
```

```python
'La palabra "Python" hace referecia a un lenguaje de programación'
```

### Nombrado de variables

Un identificador **válido** es una secuencia de caracteres (se distingue entre mayúsculas y minúsculas) que debe cumplir una serie de **condiciones**:

* Tiene que **empezar por una letra o el símbolo _**
* No puede contener espacios
* Aunque puede contener caractéres especiales, como la ñ, se recomienda no utilizarlos (estándar en programación)

```python
unNumero = 1
_un_numero = 0.123

_1mensaje = "Hola"
siguiente_mensaje_numero2 = "Mundo"
```

- Un literal (un valor constante, como un número o un texto)
- Una variable
- Una operación entre literales o variables

### Operadores aritméticos

Acabamos de utilizar operaciones para sumar y multiplicar valores numéricos. Estas operaciones aritméticas son comunes en su mayor parte a todos los lenguajes de programación:

Sumas:

```python
2 + 3
```

Restas:

```python
2 - 3
```

También podemos declarar literales que sean números negativos:

```python
-6
```

Multiplicaciones:

```python
2 * 3
```

Divisiones:

```python
5 / 2
```

Las divisiones siempre devuelven decimales, aunque sea .0

```python
4 / 2
```

Si sabemos que el resultado es entero podemos usar la doble barra para forcar a que no nos devuelva decimales:

```python
4 // 2
```

En realidad, esta operación **trunca** la parte decimal, ¡así que úsala con cautela!

```python
5 // 2
```

El módulo es una operación muy utilizada en programación, y se corresponde con el resto de la operación división. En este caso, el resto de 8 / 3 es 2:

```python
8 % 3
```

Finalmente, también podemos usar la potencia entre dos números:

```python
2 ** 3
```

### Jerarquía (precedencia) de operadores

La jerarquía natural de las operaciones es la habitual: primero potencias, luego multiplicación y división, y por último sumas y restas. Se puede utilizar los paréntesis para cambiar el orden natural:

Ejemplo: $3 * 2 ^ 3 + 5 = 3 * 8 + 5 = 24 + 5 = 29$

```python
3 * 2 ** 3 + 5
```

Ejemplo: $(3 * 2)^{(3 + 5)} = 6 ^ 8 = 1679616$

```python
(3 * 2) ** (3 + 5)
```

## Operadores sobre string

Anteriormente hemos utilizado operadores aritméticos para operar con números (suma, resta, multiplicación, ...). 
Algunos de estos operadores están __sobrecargados__ para permitir operar sobre strings, siendo diferente el efecto de estos operadores para números o para textos.

En concreto, se suelen utilizar dos operadores sobre strings:

- **suma**: permite concatenar textos, es decir, "pegar" un texto con otro
- **multiplicación**: permite repetir un texto tantas veces como indiquemos mediante un número entero

Unión de dos textos, un nombre y un apellido:

```python
nombre = "Javier"
apellido = "Cózar"

nombre_completo = nombre + " " + apellido
```

```python
nombre_completo
```

Repetición del texto "-" veinte veces:

```python
"-" * 20
```

Existe la misma jerarquía entre los operadores `+` y `*`, vista anteriormente.

En este ejemplo primero aplica la repetición de b, y despues concatena a con el resultado:

```python
"a" + "b" * 4
```

En este otro ejemplo primero aplica la concatenación, y despues la repetición:

```python
("a" + "b") * 4
```

### Formateo de strings

En ocasiones es muy interesante generar strings que expliquen algún elemento que hemos obtenido mediante nuestro programa. En este caso nos interesa combinar texto con el valor almacenado en una variable. Por ejemplo, podemos usar el operador de concatenación para lograr este efecto:

```python
compra = 123.5
impuestos = compra - compra / 1.21

"Mi compra por valor de " + str(compra) + " tiene " + str(impuestos) + " de impuestos"
```

Sin embargo es bastante engorroso:
- Tenemos que hacer **castings a strings**, ya que la concatenación de strings solo es válida entre expresiones de tipo `str`
- Tenemos que usar el **operador de concatenación**, que resta legibilidad
- si queremos **formatear** los valores a, por ejemplo, 2 decimales, tenemos que convertirlo previamente usando programación más avanzada

Para ayudarnos en esta labor, existen los denominados `f-strings`. Estos strings deben ser construidos indicando la letra `f` inicialmente y antes de abrir las comillas. Posteriormente, entre llaves `{}` introduciremos la variable a insertar en el string:

```python
f"Mi compra por valor de {compra} tiene {impuestos} de impuestos"
```

Ademas de evitar la concatenación, podemos formatear los valores para que tengan, por ejemplo, 2 decimales (aplicando redondeo):

```python
f"Mi compra por valor de {compra:.2f} tiene {impuestos:.2f} de impuestos"
```

```python
"Mi compra por valor de {:.2f} tiene {:.2f} de impuestos".format(compra, impuestos)
```

Existen formateos más sofisticados, como este para los porcentajes:

```python
iva = 0.21
f"Anteriormente se ha aplicado un {iva:.0%} de IVA"
```
