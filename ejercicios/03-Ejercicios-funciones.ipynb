### Ejercicio 2

La fórmula para pasar de grados Fahrenheit a Centígrados es la siguiente:

$$
Centigrados = \frac{(Fahrenheit - 32) \cdot 5}{9} 
$$

Crear una función que reciba los grados en Fahrenheit y devuelva la temperatura en Centígrados. ¿Cuál es el equivalente a 25 grados Fahrenheit en Centígrados?

```python
def centigrados(faren):
    return ((faren - 32) * 5) / 9


centigrados(25)
```

### Ejercicio 3

Crear una función que reciba como argumentos los siguientes datos de una persona:

- El nombre
- El primer apellido
- El segundo apellido
- La edad
- Lugar de nacimiento (opcional, con valor `desconocido`)

Esta función debe devolver un f-string con el siguiente formato:

`La persona <nombre> <appellido1> <apellido2> ha nacido en el año <año_nacimiento> en <lugar de nacimiento>`.

Para calcular el año de nacimiento suponer que la edad se ha proporcionado en el año actual.

**NOTA**: elegid un nombre adecuado para esta función. Evitad llamarle `funcion`, `mifuncion` o `ejercicio3`.

```python
def Persona(nombre,apellido1,apellido2,edad,lugardenacimiento = "Desconocido"):
    aniohoy = 2025
    return f"La persona {nombre} {apellido1} {apellido2} ha nacido en el año {aniohoy - edad} en {lugardenacimiento}"
```

```python
Persona("Pablo","Ruiperez","del Rio",23,"Valencia")
```

```python
Persona("Juan Carlos","Oliver", "Sáez",27)
```

### Ejercicio 4

Escribe el siguiente código:

- Usa la función `input` para introducir dos números entero por teclado y guardarlos en una variable. Usa casting para asegurar que son enteros.

- Suma ambos números, y crea un f-string que muestre el texto "La suma de X y Y es Z", reemplazando X, Y y Z por los valores adecuados.

**NOTA**: la función `input` siempre genera el texto introducido por teclado como un string. Si queremos que sea un valor numérico debemos castearlo.

```python
try:
    x = int(input("Introduce el número 1 por teclado"))
    y = int(input("Introduce el número 2 por teclado"))
    z = x + y

    print(f"La suma de {x} y {y} es {z}")
except ValueError:
    print("Por favor, introduce solo números enteros.")

```

### Ejercicio 5

Una función puede implementarse sin utilizar la instrucción `return`. Estas funciones no devuelven ningún valor, y son útiles cuando queremos crear una función que realice alguna acción concreta (hacer un backup de una base de datos, enviar un email, reproducir una canción en nuestro ordenador, etc.). (El equivalente a las funciones tipo void en Java)

A continuación vamos a crear una función que no devuelva ningún valor. El efecto será simplemente que imprime un mensaje por pantalla.

Por ejemplo `log_date(2021, 5, 23)` debe imprimir el texto `Hoy es 2021-05-23`.
