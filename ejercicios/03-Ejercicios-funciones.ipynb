# Funciones

## Ejercicios

### Ejercicio 1

Crea una función llamada `es_divisible` que reciba dos argumentos, `x` e `y`, e indique si `x` es divisible por `y` o no (debe devolver un booleano).

Comprueba si `7893209` es divisible por 9. ¿Y por 23?

### Ejercicio 2

La fórmula para pasar de grados Fahrenheit a Centígrados es la siguiente:

$$
Centigrados = \frac{(Fahrenheit - 32) \cdot 5}{9} 
$$

Crear una función que reciba los grados en Fahrenheit y devuelva la temperatura en Centígrados. ¿Cuál es el equivalente a 25 grados Fahrenheit en Centígrados?

### Ejercicio 3

Crear una función que reciba como argumentos los siguientes datos de una persona:

- El nombre
- El primer apellido
- El segundo apellido
- La edad
- Lugar de nacimiento (opcional, con valor `desconocido`)

Esta función debe devolver un f-string con el siguiente formato:

`La persona <nombre> <appellido1> <apellido2> ha nacido en el año <año_nacimiento> en <lugar de nacimiento>`.

Para calcular el año de nacimiento suponer que la edad se ha proporcionado en el año actual.

**NOTA**: elegid un nombre adecuado para esta función. Evitad llamarle `funcion`, `mifuncion` o `ejercicio3`.

Llamar a la función con los argumentos que consideréis oportunos, pero de la siguiente manera:

1. Todos los argumentos posicionales
1. Todos los argumentos por nombre
1. Un argumento posicional y el resto por nombre
1. Todos los argumentos por nombre menos el opcional (lugar de nacimiento) que permanecerá con su valor por defecto.

### Ejercicio 4

Escribe el siguiente código:

- Usa la función `input` para introducir dos números entero por teclado y guardarlos en una variable. Usa casting para asegurar que son enteros.

- Suma ambos números, y crea un f-string que muestre el texto "La suma de X y Y es Z", reemplazando X, Y y Z por los valores adecuados.

### Ejercicio 5

Solicita dos números por teclado. Haciendo uso de la función `max`, almacena en una variable el valor máximo de ambos.

**NOTA**: la función `input` siempre genera el texto introducido por teclado como un string. Si queremos que sea un valor numérico debemos casearlo.

### Ejercicio 6

Una función puede implementarse sin utilizar la instrucción `return`. Estas funciones no devuelven ningún valor, y son útiles cuando queremos crear una función que realice alguna acción concreta (hacer un backup de una base de datos, enviar un email, reproducir una canción en nuestro ordenador, etc.).

A continuación vamos a crear una función que no devuelva ningún valor. El efecto será simplemente que imprime un mensaje por pantalla.

Crea una función llamada `log_date` que reciba como argumentos el año, mes y día actual e imprima usando la función `print` el texto `"Hoy es YYYY-MM-DD"`.

Por ejemplo `log_date(2021, 5, 23)` debe imprimir el texto `Hoy es 2021-05-23`.
