### Ejercicio 1

Dada una lista con valores numéricos, imprimir por pantalla cada uno de ellos multiplicado por dos. Para ello haz uso del **bucle for**.

```python
numeros = [9, 5, 2, 0, 6]
```

```python
i = 0
for num in numeros:
    print(f"Numero en la posición {i} multiplicado por 2: {num * 2}")
    i = i+1
```

Como resultado deberás ver lo mismo que en la siguiente celda, donde se han utilizado 5 instrucciones print.

```python
print(numeros[0] * 2)
print(numeros[1] * 2)
print(numeros[2] * 2)
print(numeros[3] * 2)
print(numeros[4] * 2)
```

### Ejercicio 2

Mostrar por pantalla los números pares del menores que 10 utilizando para ello un **bucle while**.

```python

num = 10
while num <= 10 and num > 0:
    if(num % 2 == 0):
        print(f"{num} es par")
    num = num - 1
print(f"He salido del bucle")
```

### Ejercicio 3

Calcula la media de una lista de números

```python
numeros = [3, 2, 4, 5, 7, 1, 10, 9, 2, 3]
```

```python
acumulador = 0
contadornum = 0

for num in numeros:
    contadornum = contadornum + num
    acumulador = acumulador + 1

media = contadornum / acumulador

print(f"La media aritmetica de la lista es: {media}")
```

### Ejercicio 4

Almacena en dos variables el valor máximo y mínimo de una lista de números

```python
numeros = [3, 2, 4, 5, 7, 1, 10, 9, 2, 3]
```

**Forma 1 de hacer este ejercicio con funciones predefinidas**

```python
maxnum = max(numeros)
maxnum
```

```python
minnum = min(numeros)
minnum
```

**Forma 2, con bucles**

```python
#Máximo
maxnum = numeros[0]
for num in numeros:
    if num > maxnum:
        maxnum = num
maxnum
```

```python
#Mínimo
minnum = numeros[0]
for num in numeros:
    if num < minnum:
        minnum = num
minnum
```

### Ejercicio 5

La multiplicación no es más que la suma de un número consigo mismo un determinado número de veces. Así, $4 * 3 = 4 + 4 + 4 = 12$.

Crear una función que reciba dos argumentos numéricos de tipo entero, `x` e `y`, y que devuelva la multiplicación $x \cdot y$. Para ello **solo puedes utilizar la operación suma y los bucles**.

```python
def multiplicar(x,y):
    acumulador = 0
    for i in range(y):
        acumulador = acumulador + x
    return acumulador

multiplicar(4,3)
```
