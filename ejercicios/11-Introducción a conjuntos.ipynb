# Introducción a conjuntos

## Conjuntos

Los conjuntos son **conjuntos de elementos no repetidos**. En este caso la indexación no tiene sentido, únicamente la comprobación de pertenencia de un elemento al conjunto. Para ello haremos uso de la palabra reservada `in`.

Las funciones utilizadas para gestionar un conjunto son:
* **Creación de conjuntos**: `set` o `{}`
* **Insertar elementos**: `add`
* **Eliminar elementos**: `remove`
* **Pertenencia de elementos al conjunto**: operador infijo `in`
* **Operaciones básicas de conjuntos**: unión, intersección y diferencia
* **Número de elementos en el conjunto**: `len`
* Operaciones sobre **iterables**: all, any, max, min y sum **(igual que para listas)**

<div class="alert alert-block alert-info">
<i class="fa fa-info-circle" aria-hidden="true"></i>
Se puede almacenar cualquier elemento de los tipos básicos (int, float, bool y str).
</div>

### Creación de conjuntos

Podemos utilizar un conjunto, por ejemplo, para recordar los paises que hemos visitado en nuestros viajes:

```python
paises_visitados = {"España", "Francia", "Italia", "España", "España"}
# Podemos observar que los países no están repetidos
paises_visitados
```

Para crear un conjunto vacío debemos utilizar la función `set()`:

```python
set()
```

También podemos usar la función `set` para crear un conjunto a partir de una secuencia de elementos (una lista, un string, ...)

```python
set([1, 2, 2, 3])
```

```python
set("Abba")
```

### Número de elementos en un conjunto

Al igual que con las listas, podemos usar la función `len` para saber cuántos elementos tenemos en un conjunto:

```python
len(paises_visitados)
```

En muchas ocasiones, podemos utilizar la creación de conjuntos a partir de una lista para calcular elementos repetidos. Por ejemplo:

```python
personas_llamadas = ["Juan", "Maria", "Juan", "Pedro", "Juan"]
personas_llamadas
```

¿Cuántas personas diferentes tengo en mi lista de personas llamadas?

```python
len(set(personas_llamadas))
```

### Inserción de elementos

La función `add` nos permite añadir nuevos elementos a un conjunto previamente definido.

```python
paises_visitados.add("Grecia")
paises_visitados
```

Si este elemento ya pertenecía al conjunto la acción no tiene efecto:

```python
paises_visitados.add("España")
paises_visitados
```

Al igual que ocurre con las listas, un conjunto puede almacenar todo tipo de elementos:

```python
s = set()      # Creamos el conjunto vacío
s.add(1)       # Añadimos el 1
s.add("Hola")  # Añadimos el string "Hola"
s.add(2.3)     # Añadimos el flotante 2.3
s
```

¡Sin embargo, no es posible insertar listas en un conjunto!

```python
s.add([1, 2, 3])
```

Un conjunto admite todo tipo de elementos siempre que sean **hasheables**, que es una propiedad de las estructuras de datos en Python. Para nosotros, los únicos elementos hasheables serán los tipos básicos (int, float, str, bool y NoneType) así como la estructura de datos **tupla**.

### Eliminación de elementos

Para eliminar elementos de un conjunto utilizaremos la función `remove` y el elemento que queremos eliminar. Si éste pertenece al conjunto se eliminará.

```python
paises_visitados = {"España", "Francia", "Italia", "España", "España"}
paises_visitados.remove("España")
paises_visitados
```

Si el elemento no pertenece al conjunto tendremos un error:

```python
paises_visitados.remove("Estados Unidos")
paises_visitados
```

### Pertenencia de elementos

Como hemos mencionado, los conjuntos no sirven para acceder a los elementos individualmente, sino para saber si un elemento **pertenece** o no al conjunto. Esta comprobación se llama función de pertenencia y se realiza mediante la palabra reservada `in`:

```python
paises_visitados = {"España", "Francia", "Italia", "España", "España"}
```

```python
"España" in paises_visitados
```

```python
"Estados Unidos" in paises_visitados
```

<div class="alert alert-block alert-info">
<i class="fa fa-info-circle" aria-hidden="true"></i>
En realidad el operador infijo <i>in</i> se puede utilizar sobre cualquier elemento que sea iterable (como listas, tuplas o diccionarios).
    
Por ejemplo: `'b' in ['a', 'b', 'c']` devolvería `True`
</div>

### Operaciones sobre conjuntos
- **unión**: crea un conjunto que es la combinación de dos conjuntos en uno
- **intersección**: crea un conjunto que contiene los elementos en común en cada conjunto
- **diferencia**: crea un conjunto con los elementos diferentes de cada conjunto

```python
paises_visitados_juan = {"España", "Francia", "Italia"}
paises_visitados_maria = {"Italia", "Estados Unidos"}
```

```python
paises_visitados_juan.union(paises_visitados_maria)
```

```python
paises_visitados_juan.intersection(paises_visitados_maria)
```

```python
paises_visitados_juan.difference(paises_visitados_maria)
```

### Uso de tuplas en conjuntos

Como hemos mencionado, uno de los potenciales usos de las tuplas es que se pueden utilizar para representar secuencias de elementos e insertarlas en conjuntos (son estructuras de datos **hasheables**).

Un ejemplo en el que pueden resultar de gran utilidad es un conjunto que representa las contraseñas que hemos utilizado en un candado numérico (de 4 cifras):

```python
passwords = {
    (1, 2, 3, 4),
    (0, 0, 0, 0),
    (2, 4, 8, 1),
    (7, 3, 4, 5),
}
```

```python
passwords
```

Ahora podemos comprobar si una contraseña la hemos utilizado ya o no:

```python
(1, 2, 3, 4) in passwords
```

```python
(1, 8, 2, 4) in passwords
```
