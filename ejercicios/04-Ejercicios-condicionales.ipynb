### Ejercicio 2

Dado un número, identificar si éste es divisible por 2, por 3, o si es divisible por 2 y 3 simultáneamente. Para imprimir por pantalla uno de los siguientes mensajes:
- "El número X es divisible por 2"
- "El número X es divisible por 3"
- "El número X es divisible por 2 y por 3"
- "El número X no es divisible ni por 2 ni por 3"

```python
x = 24

# Comprobar divisores
div2 = x % 2 == 0
div3 = x % 3 == 0

if div2 and div3:
    print(f"{x} es divisible por 2 y por 3.")
elif not div2 and not div3:
    print(f"{x} no es divisible ni por 2 ni por 3.")
elif div2 and not div3:
    print(f"{x} es divisible solo por 2.")
elif div3 and not div2:
    print(f"{x} es divisible solo por 3.")


```

### Ejercicio 4

La siguiente celda genera un número aleatorio entre 0 y 10.

```python
from random import randint

numero_buscado = randint(0, 10)

```

Crea un código que pregunte por teclado el siguiente mensaje: `¿El número aleatorio es mayor o menor que 5? Introduce el símbolo > si es mayor, < si es menor o = si es igual a 5`.

Crear una sentencia condicional que nos indique por pantalla si hemos acertado.

```python
aleatorio = input("¿El número aleatorio es mayor o menor que 5? Introduce el símbolo > si es mayor, < si es menor o = si es igual a 5")
if aleatorio == "=" and 5 == numero_buscado:
    print("El número es 5")
elif aleatorio == ">" and 5 > numero_buscado:
    print("El número es mayor a 5")
elif aleatorio == "<" and 5 < numero_buscado:
    print("El número es menor a 5")
else:
    print("Fallaste. Intentalo más tarde")
```
