# Listas avanzadas

## Listas

Las listas son **secuencias de elementos** que pueden ser creadas por medio de los `[]`.

Además de crear listas podemos modificar los elementos que hay dentro de una lista, borrarlos o añadir nuevos elementos. En python toda esta funcionalidad se nos brinda de forma extremadamente cómoda y flexible, **no siendo necesario especificar el tamaño máximo de elementos** (crece según sea necesario) **ni el tipo de datos que albergará** (puede ser cualquier combinación).

Las funciones que utlizaremos para gestionar las listas son:

* **Creacón de listas**: `list` o `[]`
* **Indexación**: acceso a los elementos de la lista usando los corchetes `[]`
* **Insertar elementos al final**: `append`
* **Concatenar**: Operador `+`
* **Insertar un elemento en cualquier posición**: `insert`
* **Borrar un elemento**: `del`
* **Número de elementos en la lista**: `len`
* Operaciones sobre **iterables**: all, any, max, min y sum

### Creación de listas

Una lista se puede inicializar de dos formas:

- Usando la función built-in list
- Usando los corchetes

La ventaja de la segunda alternativa es que podemos inicializar la lista con una secuencia de valores. En cambio, la función list admite un argumento, pudiendo generar una lista a partir de cualquier secuencia de valores (función `range`, un string, y otros elementos de Python que se puedan _iterar_).

<div class="alert alert-block alert-info">
<i class="fa fa-info-circle" aria-hidden="true"></i> <strong>Nota</strong>: La función range permite generar una secuencia de elementos. En su formato de uso más básico, adamite un número entero como argumento, y genera una secuencia de números enteros desde el 0 hasta el número indicado de forma exclusiva. Por ejemplo, range(3) genera la secuencia 0,1,2 (el 3 no se incluye).
</div>  

````python
# lista vacía
list()
````

````python
# lista con los números del 0 al 9
list(range(10))
````

````python
# lista vacía
[]
````

````python
# lista con números específicos
[2, 5, 7, 1]
````

#### range

La función `range` es una función **built-in** que en realidad es mucho más potente. Además de generar secuencias de números enteros consecutivos comenzando desde el 0, permite personalizar mucho más esta secuencia. Admite hasta 3 argumentos: `(valor inicial inclusive, valor final exclusive, incremento)`:

````python
list(range(5, 12, 2))
````

El incremento puede ser **positivo** o **negativo**, pudiendo generar secuencias de números ascendentes o descendentes:

````python
list(range(10, 6, -1))
````

Si la secuencia no tiene sentido se genera una secuencia vacía (sin elementos). Por ejemplo, si indicamos `Los números que comienzan en 10, terminan en 15 con saltos de -1 en -1` esto nos genera una secuencia vacía de elementos:

````python
list(range(10, 15, -1))
````

Algunos argumentos son **opcionales** (se pueden omitir). Si solo se especifican 2, corresponden con el valor inicial y final, y se supone que el incremento es +1

````python
# Algunos argumentos se pueden omitir. Si solo se especifican 2, corresponden
# con el valor inicial y final, y se supone el incremento = +1
print( list(range(1, 3)) )  # 1 y 2

# Por ello, si se empieza en 3 y se termina en 1 con incrementos de +1, la secuencia es vacía
print( list(range(3, 1)) )  # Secuencia vacía
````

Si solo se especifica un argumento, éste es el valor final, siendo el valor inicial 0, y el incremento +1

````python
# Si solo se especifica un argumento, éste es el valor final, siendo el valor
# inicial 0, y el incremento +1
print( list(range(5)) )  # 0, 1, 2, 3 y 4
````

### Indexado: Acceso a los elementos de una lista

Podemos acceder a un elemento de una lista indicando el índice de dicho elemento entre corchetes.

````python
edades = [25, 18, 50, 67]
edad = edades[2]
print(f"La tercera persona tiene una edad de {edad} años")
````

Una forma cómoda de aceder a los últimos elementos de la lista es utilizar índices negativos, para indicar que empezamos a contar desde el final.

````python
edad = edades[-1]
print(f"La última persona tiene una edad de {edad} años")
````

### Indexación avanzada

Se puede acceder a los elementos de una lista individualmente mediante el uso de corchetes y el índice o posición del elemento al que queremos acceder. Sin embargo, también podemos recuperar un subconjunto de elementos especificando un rango de índices separados por dos puntos `:`. A esta operación se le denomina **slicing**.

En realidad, podemos especificar exactamente los mismos argumentos que en la función range: posición inicial, posición final y la longitud de los saltos. La única diferencia es de sintáxis, ya que que estos valores están separados por `:` en lugar de por comas. 

Igualmente, los índices pueden ser positivos, negativos o incluso no especificarlos, lo que significa que seleccionamos todos los elementos desde/hasta el principio/final de la lista.

````python
lista = list("ABCDEFGHIJ")
lista
````

````python
# Obtiene los elementos desde el índice 0 (inclusive) hasta el índice 5 (exclusive)
# con saltos de dos en dos posiciones
print(lista[0:5:2])
````

````python
# Obtiene los elementos desde 0 (inclusive) hasta 3 (exclusive)
print(lista[:3])
````

````python
# Obtiene los elementos desde 5 (inclusive) hasta el final
print(lista[5:])
````

````python
# Obtiene la lista inversa
# Podemos controlar los saltos entre posiciones consecutivas con un tercer elemento
print(lista[::-1])
````

````python
# Obtiene la lista quitando el primer y último elemento
print(lista[1:-1])
````

````python
# Igual que antes pero en orden inverso
print(lista[-2:1:-1])
````

````python
lista[1::3]
````

### Añadir elementos

La función `append` permite añadir elementos **al final** de una lista. Esta función modifica la lista original.

````python
# Podemos extender la lista añadiendole nuevos elementos por el final
edades = [25, 18, 50, 67]
edades.append(90)

edades
````

<div class="alert alert-block alert-warning">
<i class="fa fa-exclamation-circle" aria-hidden="true"></i>
Como una lista puede albergar cualquier elemento, puede albergar una lista también. Si ejecutamos lista.append([1, 2, 3]) no introduce los elementos 1, 2 y 3, sino que introduce un elemento (lista) que contiene los elementos 1, 2 y 3.
</div>

````python
lista = [1, 2, 3]
lista.append([1, 2, 3])

lista
````

### Concatenación de listas

Podemos crear una nueva lista a partir de dos listas concatenándolas con el operador `+`, similar a como lo utilizabamos con los strings.

````python
temperaturas_enero = [9, 5, 10, 11]
temperaturas_febrero = [11, 9, 15, 10]
temperaturas_marzo = [15, 19, 16, 20]

temperaturas_q1 = temperaturas_enero + temperaturas_febrero + temperaturas_marzo
temperaturas_q1
````

````python
# Las listas iniciales no han sido modificadas
print(temperaturas_enero)
print(temperaturas_febrero)
print(temperaturas_marzo)
````

### Inserción de elementos

Existe la función insert para insertar un elemento en una posición determinada.

Por ejemplo, podemos insertar la edad 65 en la posición 1 (será el segundo elemento de la lista):

````python
edades = [25, 19, 20, 35]
edades.insert(1, 65)

edades
````

Si queremos insertar un elemento al final de la lista normalmente usaremos `append`. Por el contrario, si queremos insertar un elemento al principio de una lista podemos usar `insert` especificando el índice 0:

````python
edades = [25, 19, 20, 35]

edades.insert(0, 18)  # la lista se modifica

edades
````

También podemos usar índices negativos. Pero cuidado, el índice negativo hace referencia al tamaño de la lista **antes** de realizar la inserción:

````python
edades = [25, 19, 20, 35]

# la edad 65 quedará almacenada en la penúltima posición ya que -1 es el índice 3
edades.insert(-1, 65)

edades
````

### Modificación de elementos

Para modificar elementos ya existentes dentro de la lista basta con utilizar el operador de indexación y el operador de asignación.

````python
# actualización de valores
edades[0] = 60

edades
````

### Borrado de elementos

Permite borrar elementos de la lista. Para ello se indexa los elementos que queremos borrar, y se precede la palabra reservada `del`

````python
edades = [25, 19, 20, 35]
edades[2]
````

````python
# borramos la edad de la tercera persona
del edades[2]
edades
````

### Tamaño de la lista

La función `len` devuelve el número de elementos que tiene una lista

````python
edades = [25, 19, 20, 35]
len(edades)
````

Es útil para acceder al último elemento, aunque lo habitual en estos casos es usar índices negativos:

````python
edades[len(edades)-1]
````

````python
edades[-1]
````

También es útil para iterar sobre los índices de una lista en lugar de sus valores. Esto se consigue combinando las funciones `range` y `len`, y puede ser necesario cuando queremos acceder a los elementos varias listas simultáneamente:

````python
spanish = ["uno", "dos", "tres"]
english = ["one", "two", "three"]

for i in range(len(spanish)):
    print(f"{spanish[i]}-{english[i]}")
````

### Operaciones sobre iterables

Un elemento **iterable** representa una secuencia de elementos: por ejemplo, las listas son elementos iterables, al igual que los strings, o estructuras de datos que veremos más tarde como los conjuntos, las tuplas o los diccionarios.

Existen funciones **built-in** que operan sobre iterables que nos pueden resultar de gran utilidad:

#### all

Devuelve `True` si todos los elementos son `True`


````python
edades = [25, 19, 20, 35, 60]
mayor_de_edad = []
for edad in edades:
    mayor_de_edad.append(edad >= 18)
mayor_de_edad
````

````python
if all(mayor_de_edad):
    print("Todos son mayor de edad")
else:
    print("Al menos hay una persona menor de edad")
````

#### any

Devuelve `True` si algún elemento es `True`

````python
edades = [25, 19, 20, 35, 16]
menor_de_edad = []
for edad in edades:
    menor_de_edad.append(edad < 18)
menor_de_edad
````

````python
if any(menor_de_edad):
    print("Al menos hay una persona menor de edad")
else:
    print("Todos son mayor de edad")
````

#### max, min y sum

Devuelven el valor máximo, minimo y la suma de la lista

````python
edades = [25, 19, 20, 35, 60]
````

````python
# La persona más mayor
max(edades)
````

````python
# La persona más joven
min(edades)
````

````python
# La edad media de todas las personas
sum(edades) / len(edades)
````

### Listas anidadas

Como mencionamos, los elementos de las listas pueden ser cualquier cosa, incluido otras listas. El siguiente ejemplo representa una tabla de datos, con dos filas y dos columnas:

col1 | col2 
--|--
1 | 2
3 | 4
5 | 6

Podemos representar una tabla como una lista que contiene filas, cada una de ellas conteniendo a su vez otra lista que contiene los valores de las columnas:

````python
tabla = [[1, 2], [3, 4], [5, 6]]
tabla
````

Podemos ver esta representación de una forma más legible usando los saltos de línea:

````python
tabla = [
    [1, 2],
    [3, 4],
    [5, 6]
]
tabla
````

Podemos iterar cada elemento de la tabla con dos bucles anidados, uno para recorrer las filas y otro para recorrer las columnas de cada fila:

````python
for fila in tabla:
    print(f"La fila completa es {fila}")
    for valor_columna in fila:
        print(f"El valor de esta columna es: {valor_columna}")
````

# Creación avanzada de listas (comprehension lists)

Es una forma de construir una lista en una sola instrucción combinando la generalidad de un bucle `for`. Esto permite crear listas con un contenido más flexible que usando por ejemplo la función `range`. Sigue la misma sintaxis que un bucle for, pero combinado con la definición de una lista (corchetes).

A modo de ejemplo, la siguiente _comprehension list_ crea una lista del 0 al 10 (el último de modo exclusive).

````python
[i**2 for i in range(10) if i**3 % 2 == 0]
````

En el siguiente _comprehension list_ se crea una lista del 0 al 10 (el último de modo exclusive) con solo los números pares.

````python
[
    i
    for i in range(10)
    if i % 2 == 0
]
````

Un **comprehension list** se divide en 3 partes, la _iteración_, en _contenido_ y el _filtrado_ (sección opcional). Es recomendable hacerlo en 3 fases:

1. Escribir la parte correspondiente a la iteración (`for`):

```
[
    for i in range(10)
]
```

2. Escribir el contenido de la lista 

```
[
    i
    for i in range(10)
]
```

3. Y finalmente escribir la condición de filtrado si lo consideramos necesario:

```
[
    i
    for i in range(10)
    if i % 2 == 0
]
```

````python
[
    i ** 2
    for i in range(10)
    if i % 2 == 1
]
````

Además de ser una forma mucho más **limpia**, **legible** y **compacta** de especificar el contenido de una lista, es mucho más **eficiente** que usando un bucle for y la función `append` o `insert` (varias órdenes de magnitud).
