# Introducción a tuplas

## Tuplas

Las tuplas son secuencias de elementos, de forma similar a las listas, pero que son **inmutables**. Esto significa que una tupla no se puede modificar una vez ésta ha sido creada.

Una de las ventajas de las tuplas, más allá de la definición conceptual, es que son estructuras de datos **hasheables**. Esto significa que **permite que estos elementos puedan ser utilizados con otras estructuras de datos como conjuntos o diccionarios**.


Al ser una especialización de las listas, tanto la **indexación** como las funciones descritas anteriormente para las listas sirven para las tuplas. La **excepción** son las funciones que modifican la lista (¡ya que las tuplas son inmutables!). Éstas son las funciones `append`, `insert` y `del`. Las funciones que trabajan con **iterables** también se pueden aplicar a tuplas: all, any, max, min y sum.

Por lo tanto, a modo de índice, éstas son las funciones que usaremos para gestionar las tuplas:
* **Creacón de tuplas**: `tuple` o `()`
* **Indexación**: acceso a los elementos de la tupla usando los corchetes `[]` **(igual que para listas)**.
* **Concatenar**: Operador `+` **(igual que para listas)**
* **Número de elementos en la tupla**: `len` **(igual que para listas)**
* Operaciones sobre **iterables**: all, any, max, min y sum **(igual que para listas)**


Conceptualmente, una tupla debe utilizarse en lugar de una lista cuando los elementos dentro de ella no se deben modificar. Por ejemplo, una persona no cambia de nombre, por lo tanto su nombre y apellidos pueden ser representados en una tupla y asignar el valor a una variable llamada persona.

En resumen, si no necesitamos modificar la secuencia de valores (ni borrar o ampliar el número de elementos), es mejor usar tuplas en vez de listas. En caso contrario, las listas son más flexibles y, por lo tanto, preferibles.

```python
persona = ("Juan", "Pérez", "López")
persona
```

Si queremos crear una tupla vacía podemos usar `tuple()` o `()`:

```python
tuple()
```

La función tuple también permite crear una tupla a partir de una secuencia de elementos:

```python
tuple([1, 2, 3])
```

```python
tuple("abcd")
```

### Indexación

Similar a las listas:

```python
persona = ("Juan", "Pérez", "López")
```

```python
# obtenemos el nombre
persona[0]
```

```python
# obtenemos los apellidos
persona[1:]
```

### Concatenación

Similar a las listas:

```python
nombre = ("Juan", )
apellidos = ("Pérez", "López")

persona = nombre + apellidos
persona
```

### Número de elementos

Similar a las listas:

```python
len(persona)
```

### Iterables

Similar a las listas:

```python
temperaturas = (10, 21, 14, 29, 34)
```

```python
# suma de todas las temperaturas
sum(temperaturas)
```

```python
# temperatura máxima
max(temperaturas)
```

```python
# temperatura mínima
min(temperaturas)
```

```python
temperaturas_frias = (True, False, True, False, False)
```

```python
is_winter = all(temperaturas_frias)
is_winter
```

```python
is_spring = any(temperaturas_frias)
is_spring
```

### Desestructuración de elementos

Python nos permite asignar una secuencia de valores a una secuencia de variables, facilitando la escritura y la lectura del código.

Por ejemplo, supongamos que tenemos una tupla con el nombre y los dos apellidos de una persona:

```python
persona = ("Juan", "Pérez", "López")
```

```python
persona
```

Podemos asignar directamente en variables independientes cada uno de los valores de esta tupla sin necesidad de estar indexando los 3 valores:

```python
# indexando valores
nombre = persona[0]
apellido1 = persona[1]
apellido2 = persona[2]

f"{apellido1} {apellido2}, {nombre}"
```

```python
# usando la desestructuración
nombre, apellido1, apellido2 = persona

f"{apellido1} {apellido2}, {nombre}"
```
