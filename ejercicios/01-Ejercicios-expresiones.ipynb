### Ejercicio 3

La fórmula para pasar de grados Fahrenheit a Centígrados es la siguiente:

$$
Centigrados = \frac{(Fahrenheit - 32) \cdot 5}{9} 
$$

Sabemos que el punto de fusión de el aluminio es 1220,54 grados Fahrenheit. ¿A qué temperatura se funde el aluminio en grados Centígrados?

```python
aluminioFahrenheit = 1220.54
aluminioCelsius = ((aluminioFahrenheit - 32) * 5) / 9

print(f"La temperatura en Celsius de fusión del aluminio es: {aluminioCelsius:.2f}" )
```

### Ejercicio 4

Sabemos cuántos años han vivido las mujeres de una determinada familia:

- Ana: 92
- María: 97
- Belén: 89
- Amparo: 87
- Rosario: 90

¿Cuál es la edad media de las mujeres de esta familia?

```python
anaedad = 92
mariaedad = 97
belenedad = 89
amparoedad = 87
rosarioedad = 90

mediaedad = (anaedad + mariaedad + belenedad + amparoedad + rosarioedad)/5

print(f"La media de edad es: {mediaedad:.2f}")
```

### Ejercicio 6

Nos interesa saber la superficie de la tierra (perímetro), sabiendo que el radio de la Tierra es 6.371Km y que la fórmula para calcular el perímetro es:

$$
perimetro = 2 \cdot \pi \cdot radio
$$

Calcula el perímetro de la tierra usando las variables `radio` y `pi`.

```python
radio = 6371
pi = 3.14

perimetro = 2 * pi * radio

print(f"El perímetro de la {perimetro:.2f} metros")

```
