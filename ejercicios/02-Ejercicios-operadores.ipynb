### Ejercicio 1

Declara una variable llamada `unidades` y otra llamada `precio`. Asegúrate que la primera es de tipo entero y la segunda de tipo flotante usando el casting.

Multiplica ambas variables y comprueba que el tipo del resultado es de tipo flotante.

```python
# Declaración de variables con casting
unidades = int(5)  # Entero
precio = float(9.95)  # Flotante

# Multiplicación
total = unidades * precio

# Verificación del tipo de dato
print("Resultado:", total)
print("Tipo de dato del resultado:", type(total))
```

### Ejercicio 2

Supongamos que hemos leido un documento que contiene el precio de un producto. Queremos multiplicarlo por `1.21` para aplicarle el IVA y almacenarlo en dicho documento de nuevo.

Sin embargo, este documento solo permite almacenar la información en formato string. Por lo tanto:

1. El número obtenido es de tipo `str`. Transforma este número a flotante (el número se encuentra almacenado en la variable `precio`)
2. Aplica la operación de multiplicación
3. Transforma el resultado a tipo `str`, para que pueda ser utilizado de nuevo en el documento

```python
iva = 0.21
preciosinivastr = "100.00"
preciofloat = float(preciosinivastr)
preciofinalfloat = preciofloat * (1 + iva)
preciofinalstr = str(preciofinalfloat)
print(f"{preciofinalstr}")
```
