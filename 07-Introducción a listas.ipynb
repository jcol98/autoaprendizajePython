# Introducción a listas

## Listas

Imagina que queremos capturar la edad de diferentes personas:

```python
edad_p1 = 25
edad_p2 = 53
edad_p3 = 45
edad_p4 = 17
edad_p5 = 61
```

¿Y si en lugar de tener las edades de 5 personas tenemos más de 20 personas? ¿O más de 100? Tendríamos que declarar ese mismo número de variables para almacenar cada una de estas edades.

Existen estructuras de datos ideadas para representar colecciones de datos. Una de las principales es la **lista**.

Las listas son **secuencias de elementos** (similar a los arrays en otros lenguajes de programación), puendo ser cada elemento de diferente tipo (enteros, flotantes, strings, booleanos, o incluso otras listas).

Para crear una lista podemos usar los corchetes, indicando el valor de cada elemento separado por comas:

```python
edades = [25, 53, 45, 17, 61]
edades
```

```python
edades[0]
edades[1]
edades[2]
edades[3]
edades[4]
```

Los elementos de una lista pueden ser de diferente tipo:

```python
numeros = [1, "dos", 3, "four", 5.0]
numeros
```

En ocasiones puede ser más legible usar saltos de línea para mostrar cada elemento en una línea de la celda diferente, pudiendo usar comentarios para describir cada uno de ellos:

```python
# Como hemos mencionado, los elementos de una lista pueden ser de diferente tipo:
numeros = [
    1,
    "dos",  # en ocasiones recibimos los números en formato texto
    3,
    "four", # e incluso en otros idiomas
    5.0  # aunque recibamos números flotantes, todos deben ser casteables a enteros
]
numeros
```

De hecho, esto también sucede así cuando usamos paréntesis. Es decir, Python sabe que empezamos a escribir un trozo de código que comienza por `(`, por lo que hasta que no encuentre un `)` interpreta que todo pertenece al mismo trozo, pudiendo escribir espacios y saltos de línea. Por ejemplo, imaginad la declaración de un string muy largo, podríamos trocearlo de esta manera:

```python
mensaje_largo = "Primera línea de un mensaje " + "que puede ser troceado en varias partes" + "usando la concatenación de strings"
mensaje_largo
```

Podemos usar los paréntesis para que cada sección del string se escriba en una sola línea:

```python
mensaje_largo = (
    "Primera línea de un mensaje " +
    "que puede ser troceado en varias partes" + 
    "usando la concatenación de strings"
)
mensaje_largo
```

Además de poder almacenar elementos de diferentes tipos básicos, ¡una lista puede incluso almacenar otras listas!

```python
# todas las llamadas del mes de mayo de 2020
llamadas = [
    ["666112233", "666112244"],  # llamadas realizadas el día 2020-05-01
    ["666112244"],  # llamadas realizadas el día 2020-05-02
    ["666112233", "666112211", "666112222"],  # llamadas realizadas el día 2020-05-03
]
llamadas
```

### Acceso a la información

Hemos visto que una lista permite almacenar la información como una secuencia de elementos. A continuación vamos a ver cómo acceder a dicha información.

Los elementos almacenados en una lista tienen un orden (primera posición, segunda posición, ..., hasta la última posición). De esta manera podríamos decir:

_Dame la edad de la tercera persona almacenada en la lista `edades`_:

<div class="alert alert-block alert-info">
<i class="fa fa-info-circle" aria-hidden="true"></i>
¡Las posiciones de los elementos de una lista se empiezan a contar desde el 0!
</div>

```python
edades
```

```python
edades[1]
```

Como podemos ver, para acceder a la información almacenada en una lista basta con especificar entre corchetes la posición del elemento al que queremos acceder, comenzando desde el 0 (primera posición), hasta $N-1$ (última posición), donde $N$ es el número total de elementos en la lista. Esta operacción de acceso a la información especificando una posición (llamada **índice**) se denomina **indexación**.

```python
# primera posición
edades[0]
```

```python
# última posición
edades[4]
```

Para acceder a la última posición debemos saber el número de elementos en la lista **y restarle 1**. Python nos proporciona una función que nos permite saber cuántos elementos tiene una lista sin necesidad de estar contándolos a mano. Esta función es `len` (viene de _length_):

```python
len(edades)
```

```python
# podemos acceder al último elemento usando esta función
edades[len(edades) - 1]
```

Con el objetivo de mejorar la usabilidad, e incluso la legibilidad, Python permite usar **posiciones negativas**. Esto es útil para empezar a contar en **orden inverso**. Es decir, -1 significa el último elemento, -2 el penúltimo, y así sucesivamente:

```python
# último elemento
edades[-1]
```

```python
# tercer elemento empezando por el final
edades[-3]
```

¿Y qué ocurre si accedemos a una posición en la que hay otra lista? Pues podemos volver a usar la indexación para acceder a la información de esta nueva lista.

```python
# lista que contiene otras listas
llamadas
```

```python
# lista almacenada en la tercera posición (índice 2)
llamadas[2]
```

```python
# segunda llamada especificada en la tercera lista
llamadas[2][1]
```

¿Y qué sucede si especificamos una posición **que no existe**? ¡Python nos dará un **error** al ejecutar esta celda!

```python
edades[5]
```

```python
edades[-6]
```
