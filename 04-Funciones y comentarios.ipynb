# Funciones y comentarios

## Funciones

Imaginemos que tenemos el siguiente código en una celda:

```python
year = 2021
month = 5
day = 17

date = f"{year}/{month:02d}/{day:02d}"
date
```

Cada vez que dispongamos de una fecha en las variables `year`, `month` y `date` y necesitemos formatearlas en un string con el formato "YYYY-MM-DD" necesitaremos replicar el código de la celda anterior a lo largo de toda la libreta.

¿Y si más tarde decidimos modificar el formato para que sea "DD-MM-YYYY"? Tendríamos que recorrer toda la libreta y modificar todas las celdas de código que formateasen la fecha.

Con el fin de simplificar esta labor y generar un código más limpio, legible y evitando repetir código, se crearon las funciones. Éstas permiten nombrar a bloques de código para ser reutilizados en múltiples partes de nuestro programa, escribiendo el código en una única celda.

La principal utilidad de las funciones es:

1. Evitar tener que repetir código constantemente
2. Hacer el código más legible

```python
def get_date(year, month, day):
    return f"{day:02d}/{month:02d}/{year}"
```

```python
get_date(2021, 5, 17)
```

```python
get_date(2021, 7, 5)
```

Si ahora queremos cambiar el formato en el que se representan las fechas, solo tenemos que modificar la función `get_date`.

Una **función** es un bloque que parte de unas entradas (argumentos), que son utilizados el código que ejecuta internamente para devolver una salida o resultado. Por ejemplo, la función `suma(x, y)` sería una función que recibe dos números `x` e `y`, y devuelve como salida la suma de ambos. El uso de una función se realiza utilizando su nombre (`suma`) e introduciendo la lista de variables de entrada (**argumentos**), separadas por comas, entre paréntesis: por ejemplo al ejecutar `suma(2,3)` devolvería 5.

Las funciones se crean usando la palabra reservada `def` seguido del **nombre** de la función:

- Para que el nombre sea válido, debe cumplir las mismas condiciones que el nombrado de variables.
- Tras el nombre de la función se abren **paréntesis** y se especifican los argumentos que se van a utilizar separados por comas (si no tiene argumentos, tan solo los paréntesis).

Dentro de la función podemos poner el código que queramos, no hay límites en tamaño ni número de líneas. Este código debe estar **indentado**. Además, deberemos usar la palabra reservada `return` para indicar que la función ha terminado de ejecutarse y queremos que devuelva el valor calculado.

A continuación vamos a declarar una función que sume dos números entre sí:

```python
def suma(x, y):
    resultado = x + y
    return resultado
```

Si ahora queremos hacer uso de esta función basta con indicar su nombre y, entre paréntesis, los valores de los argumentos declarados.

**Es importante** ejecutar la celda anterior para que la función se registre en la libreta, si no no podremos hacer uso de ella.

```python
z = suma(2, 3)
z
```

El resultado de ejecutar la función devuelve un valor que puede ser almacenado en una variable, al igual que otros literales o expresiones:

```python
peras = 10
manzanas = 3

piezas_fruta = suma(peras, manzanas)
```

```python
piezas_fruta
```

A continuación vamos a declarar una función que, dado el precio base de un producto, calcule el precio final calculando el IVA:

```python
def suma_impuestos(precio):
    return precio * 1.21
```

Y ahora hacemos uso de a función para calcular precios con impuestos:

```python
suma_impuestos(100)
```

El uso de funciones mejora la legibilidad del código (¡si usamos buenos nombres para las funciones!) y evita tener que duplicar código por toda la libreta (hasta ahora nuestro código es muy sencillo, pero cuando se complica las funciones se vuelven muy útiles).

Existen diferentes opciones para especificar los argumentos de las funciones, dotándolas de mayor expresividad y potencial.

### Argumentos opcionales

La función anterior aplica el 21% de IVA a cualquier precio pasado como argumento. Esto es útil si el IVA es fijo, pero si queremos usar la función para aplicar el IVA a todo tipo de productos (alimentación básica, ocio, etc.) necesitamos más flexibilidad.

Python nos permite declarar parámetros con un valor por defecto indicado. Estos parámetros se llaman **opcionales**, pues al declararse con un valor por defecto no es necesario especificarlo si no queremos, pero nos da la opción por si queremos cambiar dicho valor.

En este ejemplo vamos a aplicar el iva a diferentes productos, cada uno de ellos con un IVA diferente asociado:

```python
def suma_impuestos(precio, iva=21):
    return precio * (1 + iva / 100)
```

```python
precio_carne = 7.5
precio_huevos = 4.75
cervezas = 12.5
```

```python
suma_impuestos(precio_carne, 10)
```

```python
suma_impuestos(precio_huevos, 4)
```

```python
suma_impuestos(cervezas)
```

### Orden de los argumentos

Cuando se llama a una función, los argumentos pueden ser proporcionados en **orden** o por **nombre**.

Estas opciones pueden aumentar enormemente la legibilidad de nuestro código:

Ejemplo por orden:

```python
suma_impuestos(precio_carne, 10)
```

Ejemplo por nombre:

```python
suma_impuestos(precio=precio_carne, iva=10)
```

Se pueden combinar ambas formas, por orden y por nombre, pero existen unas reglas:

- Mientras declaramos argumentos por posición, hacemos referencia a todos los argumentos en orden de izquierda a derecha
- En el momento en el que se declara el primer argumento por nombre, los parámetros a continuación se declaran por nombre también

```python
suma_impuestos(iva=10, precio=precio_carne)
```

## Comentarios

Todo lenguaje de programación permite escribir comentarios. Éstos son lineas de texto explicativo que se usa para documentar el código. Esto es muy útil para explicar con vocabulario la funcionalidad de determinadas expresiones, celdas, o bloques de código y funciones.

Los comentarios son líneas que comienzan con el carácter `#`. El texto que se escribe a continuación es ignorado por python, y tiene una funcionalidad puramente documentativa. 

**NOTA**: los comentarios forman parte del lenguaje Python, por lo que las celdas en las que se incluya deben ser de tipo código.

```python
# Este texto no se ejecuta en python

# Cada línea debe comenzar con el hashtag si queremos que sea un comentario
```

También se pueden usar comentarios al final de una línea de código:

```python
resultado = 4 / 2  # el resultado siempre es un float
```

```python
resultado
```

A diferencia de otros lenguajes, Python no dispone de comentarios multilínea. Es decir, si queremos comentar múltiples líneas deberemos usar el símbolo `#` individualmente en cada una de ellas.

Existe la poibilidad de declarar strings que contengan saltos de línea, permitiendo especificar un string cómodamente como un párrafo. Mucha gente utiliza estos elementos, llamados `docstrings` para comentar bloques de texto. De hecho, el nombre `docstring` viene de utilizar estos strings para documentación del código. La forma de declarar estos strings es usando tres comillas para abrir y tres para cerrar el string:


```python
"""Esto es un docstring.

Permite indicar un text en múltiples líneas.

En ocasiones se usa para comentar múltiples líneas simultáneamente, pero hay
que tener en cuenta que esto es un string, por lo que si ejecutamos esta celda se mostrará
como salida!"""
```

En realidad, en jupyter, cualquier celda que termine con una expresión que no se asigne a ninguna variable mostrará el valor de dicha expresión como salida de la celda.

```python
# Ejemplo en el que se muestra el resultado de la operación en la celda
2 + 3
```

```python
# Al asignar el valor a una variable ya no se muestra
x = 2 + 3
```

```python
# Si especificamos la variable, que tiene el valor 5 almacenado, se mostrará
x
```

```python
"""Si usamos un docstring pero tenemos código a continuación, el docstring no se
mostrará.
"""
x = 2 + 3
```

## Parámetros avanzados

Cuando declaramos una función de Python podemos pasar cualquier tipo de elemento, como enteros, listas y también diccionarios. Por ejemplo, podemos pasar una lista que contenga varios valores y nos devuelva la suma de los que sean únicamente pares:

```python
def suma_pares(lista):
    suma = 0
    for x in lista:
        if x % 2 == 0:
            suma += x
    return suma
```

```python
suma_pares([1, 2, 3, 4, 5])
```

Python dispone de una alternativa más elegante para pasar un número indeterminado de elementos, usando `*args`. Se expresa indicando un asterisco antes del nombre de la variable en el argumento, lo que significa que todos los argumentos posicionales que no hayan encajado con parametros anteriores se asignaran como una lista en la variable:

```python
def suma_pares(*args):
    suma = 0
    for x in args:
        if x % 2 == 0:
            suma += x
    return suma
```

```python
suma_pares(1, 2, 3, 4, 5)
```

Como vemos la diferencia está en que pasamos los valores como un número indeterminado de argumentos, no como un argumento de tipo lista.

También disponemos de otra opción para indicar un número indeterminado de argumentos, pero en este caso nombrados. Para ello, en lugar de usar un asterisco se usa doble asteriso `**kwargs`:

```python
def log_message(timestamp, severity, **data):
    print(f"{timestamp} - [{severity}]: Se ha generado el log con los siguientes campos")
    for k, v in data.items():
        print(f"{k}: {v}")
```

```python
import time
```

```python
log_message(time.time(), "ERROR", image_size=(512, 512), type="jpeg")
```

```python
log_message(time.time(), "ERROR", name="Javi", age=35)
```

### Utilidad

La utilidad de pasar una función como argumento de otra función se centra principalmente en generalizar.

Por ejemplo, imaginemos que disponemos de una función que recibe un texto y nos lo envía por email (por simplicidad, vamos a implementar simplemente que nos lo muestre por pantalla). Sin embargo, podemos personalizar el texto pasando una lista de funciones que transformarn el texto:
 

```python
def send_email(recipient, text, *args):
    for fn in args:
        text = fn(text)
    print(f"Enviando email a {recipient}...")
    print(text)
```

```python
send_email(
    "person@gmail.com",
    "soy javi, quedamos mañana a las 16:30."
)
```

```python
send_email(
    "person@gmail.com",
    "soy javi, quedamos mañana a las 16:30.",
    str.upper
)
```

```python
send_email(
    "person@gmail.com",
    "soy javi, quedamos mañana a las 16:30. saludos!",
    str.capitalize
)
```

```python
def capitalize_sentences(text):
    sentences = text.split(".")
    for i in range(len(sentences)):
        sentences[i] = sentences[i].strip().capitalize()
    return '. '.join(sentences)

send_email(
    "person@gmail.com",
    "soy javi, quedamos mañana a las 16:30. saludos!",
    capitalize_sentences,
)
```

```python
def capitalize_sentences(text):
    sentences = text.split(".")
    for i in range(len(sentences)):
        sentences[i] = sentences[i].strip().capitalize()
    return '. '.join(sentences)

def capitalize_names(text):
    names = ["Pedro", "Laura", "Javi", "María"]
    for name in names:
        text = text.replace(name.lower(), name)
    return text

send_email(
    "person@gmail.com",
    "soy javi, quedamos mañana a las 16:30. saludos!",
    capitalize_sentences,
    capitalize_names
)
```

## Funciones anónimas

Además de la definición estándar de funciones existe la posibilidad de crear funciones en una sola línea. Estas funciones se llaman `lambda`, y son funciones que ejecutan una expresión y devuelve ese resultado. Estas funciones siempre tienen su equivalencia en una declaración de función estándar, la ventaja es que necesitamos escribir mucho menos y además en una sola línea de código. Para ello indicaremos `lambda` seguido de los argumentos separados por comas (sin usar paréntesis), después los dos puntos, y el códio que represente la operación que hará la función. **Siempre** se devuelve el resultado de la expresión utilizada **sin necesidad** de usar `return` (no se pone, pero devuelve el resultado).

Se utilizan mucho cuando son argumentos de otras funciones, ya que son funciones que no necesitamos reutilizar, simplemente crear para pasar como argumento para otras.

```python
def suma(x, y):
    return x + y

suma(2, 3)
```

```python
# uso equivalente usando lambda (no tiene sentido asignarla a una variable, pero lo haremos por legibilidad)
suma = lambda x, y: x + y

suma(2, 3)
```

```python
a = 10
b = 20

# Declarando la función
def opera(x, y, fn):
    return fn(x, y)

resultado_sumar = opera(a, b, lambda x, y: x + y)
resultado_multiplicar = opera(a, b, lambda x, y: x * y)

print(f"{a} sumar con {b} = {resultado_sumar}")
print(f"{a} multiplicar con {b} = {resultado_multiplicar}")
```

En python existen funciones como `max`, `min` o `sorted` que admiten un argumento nombrado `key` que es una función. El objetivo es permitir personalizac cómo realizar la ordenación de elementos, tanto para ordenar (`sorted`) como para obtener el máximo y el mínimo valor (`max` o `min`).

En el ejemplo a continuación vamos a tener una lista de tuplas, indicando el nombre de personas y su edad:

```python
people = [
    ("Javi", 35),
    ("Pedro", 31),
    ("María", 18),
    ("Laura", 40),
    ("José", 20),
    ("Daniel", 65),
]
```

Si ordenamos `people` lo hará en base al primer elemento de la tupla (comportamiento por defecto), que resulta en un orden alfabético:

```python
sorted(people)
```

Sin embargo, podemos hacer que ordene por la edad usando el argumento `key`, proporcionando una función que recibe un elemento de la lista (las tuplas) y devuelve un valor que se usará para ordenar (el valor de la edad):

```python
sorted(people, key=lambda person: person[1])
```

O en orden inverso:

```python
sorted(people, key=lambda person: person[1], reverse=True)
```

```python
# equivalente
sorted(people, key=lambda person: -person[1])
```

Lo mismo se puede aplicar para `max` y `min`:

```python
max(people, key=lambda person: person[1])
```

```python
min(people, key=lambda person: person[1])
```
