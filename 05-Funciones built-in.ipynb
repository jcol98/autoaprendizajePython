# Funciones built-in

Las funciones built-in son funciones que Python nos da ya implementadas. A continuación se describen las que pueden resultar de mayor interés. No obstante, en este [enlace](https://docs.python.org/3/library/functions.html) se puede ver la lista completa de funciones built-in.

* **print**: Función que imprime en la salida el string pasado como argumento. Si no es un string, intentará castearlo a string.
* **abs**: Computa el valor absoluto
* **bool**: Fuerza un elemento (cast) a ser de tipo booleano
* **int**: Fuerza un elemento (cast) a ser de tipo entero
* **float**: Fuerza un elemento (cast) a ser de tipo flotante
* **str**: Fuerza un elemento (cast) a ser de tipo string
* **max** : Computa el máximo entre los números pasados como argumentos
* **min** : Computa el mínimo entre los números pasados como argumentos
* **round**: Redondea un número flotante a entero. **Utiliza el método de redondeo del banquero**
* **type**: Devuelve el tipo de la variable pasada como argumento.
* **input**: Permite solicitar un valor por teclado para, por ejemplo, signarlo a una variable
* **isinstance**: Indica si una variable o literal es del tipo especificado o no (devuelve un booleano)

#### print

```python
# imprime lo que se le pasa como argumento a modo de string
print("Hola mundo")

# Se le puede pasar más de un argumento, y los concatena usando un espacio
print("Hola", "mundo")

# Los argumentos pueden ser de diferentes tipos (los que no son String los convierte)
print(2, "más", 2, "son", 4)
```

#### abs

```python
# Función valor absoluto
abs(-1)  # Devuelve 1
```

#### Casting: bool, int, float y str

```python
# Constructores de elementos booleanos, enteros, float y strings

# bool
x = 0
print( bool(x) )   # Castea el entero 0 a booleano (False)

# int
x = 2.5
print( int(x) )    # Castea el número flotante a entero por truncamiento

# float
x = 1
print( float(x) )  # Castea el número entero a flotante

# str
x = 3.14
str(x)    # Castea el flotante a string
```

#### max y min

```python
# max y min
a = 3
b = 7
c = 6

print( max(a, b, c) )  # Imprime el valor máximo de a, b y c
print( min(b, a, c) )  # Imprime el valor mínimo de a, b y c
```

#### round

Función round. Redondea al entero siguiendo el método del banquero: 
 - Se trata de un redondeo más justo. Es un redondeo típico si la parte decimal es >.5 o <.5. Sin embargo, si es exactamente .5: si la parte entera es par, se redondea como si fuese <.5, y si es impar como si fuese > .5

```python
# Se redondea al entero más cercano
print( round(10.4) )  # 10.4 se redondea a 10
print( round(10.6) )  # 10.6 se redondea a 11
print( round(11.4) )  # 11.4 se redondea a 11
print( round(11.6) )  # 11.6 se redondea a 12

print()
# En cambio, si la parte decimal es .5 exactamente
# Si la parte entera es par, se redondea como si fuese <.5
print( round(10.5) )  # 10.5 se redondea a 10
print( round(-10.5) ) # -10.5 se redondea a 10
# Si la parte entera es impar, se redondea como si fuese >.5
print( round(11.5) )  # 11.5 se redondea a 12
print( round(-11.5) ) # -11.5 se redondea a -12
```

#### type

```python
# Por último, la función type devuelve el tipo de la variable
print( type(True) )         # Es un booleano
print( type(2) )            # Es un entero
print( type(1.0) )          # Es un flotante
print( type("Hola mundo") ) # Es un string
```

### input

`input()` solicita un valor por teclado. En jupyter notebook aparece un campo de texto en la sección de salida: se introduce el valor y se confirma pulsando intro

```python
x = input()
print("Has introducido el valor: ")
print(x)
print(type(x))

# Se puede introducir un texto informativo en el input:
x_string = input("Introduce un número entero: ")
x_int = int(x_string)
print(type(x_int))
```

### isinstance

Aunque puede resultar extraño, el segundo argumento (tipo) debe proporcionarse indicando las funciones de casting vistas anteriormente (`bool`, `int`, `float` o `str`).

En Python todo puede almacenarse en variables, incluidas las funciones.

En Python todo puede proporcionarse como argumentos a funciones, incluidas otras funciones.
