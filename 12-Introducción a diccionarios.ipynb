# Introducción a conjuntos

## Diccionarios

Los diccionarios son elementos que relacionan una **clave** con un **valor**. La clave tiene que ser un tipo básico o una tupla (elemento _hasheable_). Las siguientes funciones son las utilizadas para gestionar diccionarios:

* **Creacón de diccionarios**: `dict` o `{k:v}`
* **Indexación**: acceso a los elementos de la tupla usando los corchetes `[]`, donde dentro de los corchetes se especifica una clave
* **Borrar un elementos**: `del` **(igual que para listas)**
* **Número de elementos en el diccionario**: `len` **(igual que para listas)**
* Operaciones especiales: keys, values, items. Permite acceder a la lista de claves, valores, o tuplas (clave, valor) del diccionario.
* Operaciones sobre **iterables**: all, any, max, min y sum. En realidad, estas funciones no operan sobre el diccionario sino que, por defecto, operan sobre sus claves. Si queremos aplicar cualquiera de estas funciones a los valores, podemos usar la función `values` del diccionario y posteriormente la función sobre iterables.

### Creación de diccionarios

Podemos usar las llaves para crear un diccionario. La diferencia con los conjuntos es que debemos especificar cada valor dividido en dos partes, la clave (a la izquierda del símbolo `:`) y el valor (a la derecha de dicho símbolo).

```python
agenda = {"juan": 666112233, "maria": 666223344}
agenda
```

Para declarar un diccionario vacío podemos usar `dict` o `{}`. Nótese que Python lo interpreta como un diccionario vacío, no un conjunto vacío:

```python
dict()
```

```python
{}
```

Aunque también se puede utilizar `dict` para inicializar un diccionario, lo habitual es usar el formato de los corchetes, ya que es más flexible.

```python
agenda = dict(juan=666112233, maria=666223344)
agenda
```

Si disponemos de una lista de tuplas de **longitud 2**, la función `dict` es capaz de construir un diccionario donde el primer elemento de cada tupla es la clave y el segundo es el valor:

```python
personas = [("juan", 666112233), ("maria", 666223344)]

dict(personas)
```

Podemos utilizar los saltos de línea para que el código sea más legible, sobre todo cuando los valores declarados son complejos. En este caso, el valor de cada clave es otro diccionario, que representa cómo se dice el número en varios idiomas:

```python
translation = {
    1: {
        "spanish": "uno",
        "english": "one"
    },
    2: {
        "spanish": "dos",
        "english": "two"
    },
}
translation
```

### Acceso a los elementos

La indexación se realiza de la misma manera que en las listas y las tuplas, usando los corchetes. La diferencia es que aquí no tenemos posiciones, sino claves:

```python
agenda = {"juan": 666112233, "maria": 666223344}
agenda
```

```python
agenda["maria"]
```

¿Cómo se dice 2 en inglés?

```python
translation[2]["english"]
```

<div class="alert alert-block alert-warning">
<i class="fa fa-exclamation-circle" aria-hidden="true"></i> Hay que tener cuidado, si accedemos a un elemento que no existe obtendremos un error del tipo KeyError

</div>

```python
translation[2]["polish"]
```

### Añadir elementos

Para insertar elementos en un diccionario basta con usar el **operador de indexación** con la clave que deseemos insertar y el **operador de asignación** con el valor que queremos asignar a la clave.

```python
agenda["Roberto"] = 666445566
agenda
```

### Borrar elementos

El borrado de elementos se realiza igual que con las listas, utilizando la palabra reservada `del` seguida de una indexación (la que queremos borrar).

```python
del agenda["Roberto"]
agenda
```

### Número de elementos en un diccionario:

Al igual que con el resto de estructuras, usando la función `len`:

```python
len(agenda)
```

### Operaciones especiales sobre diccionarios

Se puede acceder a la lista de claves, la lista de valores, y la lista de tuplas (clave, valor) de un diccionario. Para ello, existen las funciones:
- `keys`: accede a la lista de claves
- `values`: accede a la lista de valores
- `items`: devuelve una lista de tuplas de longitud 2, siendo la clave el primer elemento y el valor el segundo

```python
agenda = {"juan": 666112233, "maria": 666223344, "roberto": 666334455}
agenda
```

En realidad, estas funciones no devuelven una lista sino que devuelven un objeto especial de forma similar a como ocurría con la función `range`.

```python
agenda.keys()
```

Estos objetos pueden usarse en bucles `for` o puede usarse la función `list` para transformarlos en listas:

```python
for nombre in agenda.keys():
    print(nombre)
```

```python
list(agenda.values())
```

```python
for telefono in agenda.values():
    print(telefono)
```

```python
for nombre, telefono in agenda.items():
    print(f"{nombre} -> {telefono}")
```
