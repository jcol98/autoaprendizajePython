# Tipos básicos

## Tipos

Cuando especificamos un literal podemos crear diferentes tipos de elementos, como por ejemplo números o textos. Estos elementos tienen asociado un **tipo**, que indica la naturaleza del literal. Podemos consultar el tipo de cualquier literal o variable en Python mediante una función llamada `type`.

<div class="alert alert-block alert-info">
<i class="fa fa-info-circle" aria-hidden="true"></i> <strong>Nota</strong>: las funciones son operaciones que realizan una acción. Por ejemplo, la función pow(2, 3) aplica la operación 2 elevado a 3. Veremos en detalle las funciones más adelante.
</div>    

Existen diferentes tipos de elementos en Python. Los principales son:
- Números enteros (`int`)
- Números reales (`float`)
- Textos (`str`)
- Valores condicionales (cierto o falso) (`bool`)
- Valor vacío (`None`)

Números enteros:

```python
1
```

```python
type(1)
```

Números reales:

```python
0.95
```

```python
type(0.95)
```

Textos:

```python
"Hola mundo"
```

```python
type("Hola mundo")
```

Valores condicionales (cierto o falso). Estos valores se representan con las palabras `True` y `False`:

```python
True
```

```python
type(True)
```

```python
False
```

```python
type(False)
```

Valor vacío. Es útil para representar ausencia de datos. Por ejemplo, si disponemos de la edad de diferentes personas identificadas por su nombre, pero para una de ellas no sabemos su edad, podemos representar esta ausencia de información mediante la palabra `None`.

```python
None
```

```python
type(None)
```

## Asignaciones

Cualquier expresión puede ser asignada a una variable. En este momento, la variable también asimila el tipo de la expresión asignada.

Por ejemplo:

```python
precio = 100 * 1.21
```

```python
type(precio)
```

A diferencia de otros lenguajes de programación, en Python los **tipos se infieren dinámicamente**, es decir, no es necesario indicar de qué tipo es una variable pero ésta tiene un tipo asociado.

Cada vez que creamos una variable por medio de la asignación (usando el símbolo `=`) la variable se crea con un tipo específico.

El operador de asignación no se limita a la creación de nuevas variables, también permite **reasignar valores a variables ya declaradas**:

Primero creamos una nueva variable:

```python
edad = 30
```

Vemos que el tipo de la variable es entero:

```python
type(edad)
```

Podemos usar el operador de asignación para cambiar el valor almacenado en la variable `edad`:

```python
edad = 50
```

```python
edad
```

```python
type(edad)
```

Vemos que el tipo no ha cambiado, solo el valor, porque en este caso también es un entero. Pero podemos asignar **cualquier valor** a una variable, incluso si ya está declarada:

```python
edad = "cincuenta"
```

```python
edad
```

```python
type(edad)
```

Esta propiedad de Python se denomina **tipado dinámico**, y no solo se refiere a la creación de variable sino también a la reasignación de valores. Esta propiedad es muy cómoda ya que no nos tenemos que preocupar de los tipos de las variables, aunque sepamos que internamente éstos existen.

### Casting

En ocasiones queremos especificar el tipo de una variable al realizar una asignación, y no depender del tipado dinámico que realiza Python. En estos casos, podemos usar las funciones `int`, `float`, `str` o `bool` para forzar que el valor de una expresión sea de un tipo concreto:

Por ejemplo, el resultado de una división siempre es un flotante:

```python
4 / 2
```

Pero podemos forzar que sea de tipo entero:

```python
int(4 / 2)
```

Se puede realizar la operación de casting a cualquier valor, ¡pero algunas de estas operaciones puede fallar si no tiene sentido!:

Por ejemplo, podemos convertir un entero a un string:

```python
str(120)
```

Y un string a un entero:

```python
int("120")
```

¡Pero solo si la conversión tiene sentido!

También existe la posibilidad de realizar una comprobación para ver si un valor está dentro de un rango:

```python
int("uno")
```

El tipo más general, y que nunca causará un problema al aplicar una operación de casting es el string. Despues, es el flotante, después el entero, y finalmente los booleanos:

```python
int(False)
```

```python
float(23)
```

```python
str(23.0)
```
