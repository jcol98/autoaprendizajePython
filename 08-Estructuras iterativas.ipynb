# Estructuras iterativas

### Bucles

Ahora imagina que queremos calcular la edad media de todas las personas.

```python
edades = [25, 53, 45, 17, 61]
```

Para calcular la media sumamos todos estos valores y dividimos por el número de personas:

```python
edad_media = (edades[0] + edades[1] + edades[2] + edades[3] + edades[4]) / len(edades)
edad_media
```

¿Y si en lugar de tener las edades de 5 personas tenemos las de 1000 personas? ¿Y un millón?

Un **bucle** nos permite repetir **un mismo código** varias veces de una forma compacta, legible y cómoda. Por ejemplo, sumar las edades de todas las personas una a una.

Existen dos alternativas para crear estos bucles:

- **for**: un bucle que repite un mismo código un número determinado de veces. En cada iteración disponemos de una variable que va cambiando de valor.
- **while**: un bucle que repite un mismo código hasta que se cumpla una condición de parada.

### for

La estructura de bucle for permite **repetir un bloque de código** tantas veces como elementos tenga una **lista**. En cada iteración sobre el bucle (ejecución del código especificado dentro del mismo), una variable tomará el valor de cada uno de los elementos de esa lista.

<div class="alert alert-block alert-info">
<i class="fa fa-info-circle" aria-hidden="true"></i> En otros lenguajes de programación es común definir los bucles for asignando un valor e incrementándolo al final de cada iteración. Esa opción no está disponible en python.
</div>

Para definir un bucle for usaremos la palabra reservada `for`, seguida del nombre de una variable que será la que tomará cada uno de los valores de la lista, a continuación la palabra reservada `in` y a continuación la lista sobre la que queremos iterar.

Al igual que con la definición de funciones y con la ramificación, usaremos los `:` y la indentación para especificar el bloque de código a ejecutar por cada elemento en la lista.

El siguiente código recorre los elementos de la lista `[1, 2, 3, 4]` e imprime cada valor por pantalla:

```python
for i in [1, 2, 3, 4]:
    print(f"Soy el número {i}")
```

Si queremos calcular la media de edad de la lista `edades` podemos sumar todas las edades una a una en una variable, y posteriormente realizar la división:

```python
suma_edades = edades[0]
suma_edades = suma_edades + edades[1]
suma_edades = suma_edades + edades[2]
suma_edades = suma_edades + edades[3]
suma_edades = suma_edades + edades[4]

media_edad = suma_edades / len(edades)
media_edad
```

Esto mismo puede ser realizado de una forma mucho más compacta con un bucle for:

```python
edades
```

```python
suma_edades = 0

for edad in edades:
    suma_edades = suma_edades + edad

media_edad = suma_edades / len(edades)
media_edad
```

Un bucle for no solo puede iterar sobre listas. En realidad lo puede hacer sobre cualquier elemento de Python que sea **iterable**, es decir, que tenga sentido _recorrerlo_. Por ejemplo, un string es una secuencia de caracteres, por lo que el bucle for puede ser utilizado para recorrer cada uno de los caracteres del texto:

```python
for c in "Hola mundo":
    print(c)
```

De la misma manera podemos acceder a la información individual de un string, es decir, usando los corchetes y un índice podemos acceder a cada uno de los caracteres individualmente:

```python
texto = "Hola mundo"
```

```python
texto[0]
```

```python
texto[-1]
```

E incluso usar la función `len`:

```python
len(texto)
```

### While

La estructura de bucle for permite **repetir un bloque de código** en función de una expresión **booleana**. Al comienzo de cada iteración se comprobará el valor de esta expresión, y se continuará ejecutando el bucle si la condición es `True`, y se terminará en caso contrario (`False`)

Para definir un bucle while usaremos la palabra reservada `while` seguida de una expresión booleana. De nuevo, usaremos los `:` y la indentación para indicar el bloque de código a ejecutar mientras que la expresión booleana se resuelva a `True`.

```python
x = 1

while x < 5:
    print(f"Soy el número {x}")
    # Sumamos 1 al valor de x. Cuando el valor de x sea 5 la expresión booleana será False
    x = x + 1
```

En el caso del bucle `while` no recorremos una lista, pero podemos utilizar una variable numérica como índice para conseguir este mismo efecto. No obstante, en esas situaciones suele ser preferible usar un bucle `for`:

```python
i = 0
suma_edades = 0

while i < len(edades):
    suma_edades = suma_edades + edades[i]
    i = i + 1

media_edad = suma_edades / len(edades)
media_edad
```

<div class="alert alert-block alert-info">
<i class="fa fa-info-circle" aria-hidden="true"></i>
<strong>Nota</strong>: La sentencia <strong>break</strong> se puede utilizar para salir de un bucle inmediatamente.
</div>

```python
x = 1
while True:
    print(f"Soy el número {x}")
    x = x + 1
    if x == 5:
        break
```

<div class="alert alert-block alert-warning">
<i class="fa fa-info-circle" aria-hidden="true"></i>
<strong>¡Cuidado!</strong>: Los bucles <strong>while</strong> son propensos a crear <strong>bucles infinitos</strong> por error, es decir, bucles que se ejecutan y nunca terminan al no satisfacer la condición de parada. En este caso podemos parar la ejecución manualmente pulsando el botón cuadrado situado al lado del botón de ejecutar celda (triángulo) o en el menú Kernel, opción Interrumpir Kernel.
</div>
