# If-else

En ocasiones es necesario implementar un código que solo se ejecute cuando se cumple una condición. Por ejemplo, si el valor de una variable es mayor que un umbral establecer la variable `color="red"`, siendo `color="blue"` en caso contrario.

Este tipo de código se denomina ramificado, y significa que en función del valor de una expresión **booleana** se ejecutará un trozo de código u otro.

En Python, la estructura de control condicional **if-else** permite **bifurcar el flujo de la ejecución de un programa** en dos alternativas. Mediante el uso de una expresión booleana (que se resuelve como `True` o `False`), perimte ejecutar uno u otro bloque, asociados a cada uno de los posibles valores booleanos.

Para indicar el comienzo de una ramificación utilizaremos la palabra reservada `if`, seguida de una expresión booleana. Si la expresión es `True` se ejecutara el bloque de código especificado a continuación del símbolo `:`:

**Nota:** el bloque de código que se especifica a continuación del if debe estar indentado por 4 espacios, al igual que en los bloques que definen las funciones.

```python
temperatura = 30

if temperatura >= 35:
    print("Hace calor")
```

Opcionalmente es posible ejecutar un bloque de código alternativo en caso de que la condición especificada no se cumpla. En este caso, se puede utilizar la palabra reservada `else` seguido de los `:` para especificar el bloque alternativo:

```python
temperatura = 40

if temperatura >= 35:
    print("Hace calor")
else:
    print("No hace calor")
```

En ocasiones se necesita ramificar el código en más de dos opciones, por ejemplo en función del día de la semana. En este caso se pueden combinar tantas veces como se necesite la instrucción `if-else`:

```python
dia_semana = "sábado"

if dia_semana == "lunes":
    print("Es lunes")
else:
    if dia_semana == "martes":
        print("Es martes")
    else:
        if dia_semana == "miércoles":
            print("Es miércoles")
        else:
            if dia_semana == "jueves":
                print("Es jueves")
            else:
                if dia_semana == "viernes":
                    print("Es viernes")
```

Como podemos ver cada nueva opción nos obliga a indentar una vez más el código. Esto hace que nuestro código pierda legibilidad y se vuelva engorroso. Como es una situación bastante habitual, Python nos proporciona la siguiente alternativa equivalente a la anterior: `elif`.

Como se puede apreciar, el código es más fácil de escribir y leer.

```python
dia_semana = "viernes"

if dia_semana == "lunes":
    print("Es lunes")
elif dia_semana == "martes":
    print("Es martes")
elif dia_semana == "miércoles":
    print("Es miércoles")
elif dia_semana == "jueves":
    print("Es jueves")
elif dia_semana == "viernes":
    print("Es viernes")
```

También es posible que necesitemos diseñar una ramificación más compleja que una simple pregunta o condición. Por ejemplo, imaginemos que estamos controlando un sistema de aire acondicionado. En función de la temperatura y de la humedad queremos establecer la potencia del aparato, por lo que necesitaremos dos condiciones:

```python
temperatura = 40
humedad = 0.1

if temperatura > 35 and humedad > 0.5:
    potencia = 1.0
elif temperatura > 35 and humedad <= 0.5:
    potencia = 0.8
elif temperatura < 35 and humedad > 0.5:
    potencia = 0.5
else:  # temperatura < 35 and humedad <= 0.5:
    potencia = 0.2

potencia
```

Otra situación en la que podemos encontrarnos ramificaciones más complejas es cuando queremos actuar condicionalmente dentro de uno de los bloques definidos por una condición. Python permite anidar tantas ramificaciones como queramos, aunque no es recomendable utilizar más de dos o tres ramificaciones anidadas (genera lo que en programación se denomina [código espagueti](https://es.wikipedia.org/wiki/C%C3%B3digo_espagueti).

```python
season = "winter"
temperatura = 25

potencia = 0
  
if season == "summer":
    if temperatura > 35:
        potencia = 1.0
    elif temperatura > 25:
        potencia = 0.5
elif season == "winter":
    if temperatura > 35:
        potencia = 0.5
    elif temperatura > 25:
        potencia = 0.25
    
potencia
```

La ramificación tiene infinitas posibilidades. Depende directamente de las expresiones booleanas y la cantidad de opciones y ramificaciones anidadas que necesitemos. Un caso de uso bastante habitual es la de discretizar una variable, que se traduce en pasar de una variable numérica a una serie de etiquetas o categorías.

Por ejemplo, dada una variable numérica de temperatura, crear una variable `temperatura_discretizada` que valga uno de los siguientes tres valores:

- "alto": si la temperatura es mayor de 35 gradis
- "medio": si la temperatura es mayor de 15 y menor de 35
- "bajo": si la temperatura es menor de 15 grados

```python
temperatura = 22

if temperatura > 35:
    temperatura_discretizada = "alto"
elif 15 < temperatura <= 35:
    temperatura_discretizada = "medio"
elif temperatura <= 15:
    temperatura_discretizada = "bajo"

temperatura_discretizada
```

La comprobación de cada una de las ramas se realiza secuencialmente en orden, de arriba a abajo. Normalmente se juega con ello para hacer las condiciones más sencillas:

```python
temperatura = 22

if temperatura > 35:
    temperatura_discretizada = "alto"
elif temperatura > 15:  # si llegamos aquí sabemos que no es >35
    temperatura_discretizada = "medio"
else:  # si llegamos aquí sabemos que no es >15
    temperatura_discretizada = "bajo"

temperatura_discretizada
```
