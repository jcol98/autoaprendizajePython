# Operadores lógicos

## Operadores lógicos

Al igual que las operaciones aritméticas permiten operar entre valores numéricos, los operadores lógicos permiten comparar elementos entre sí, devolviendo como resultado literales booleanos (`True` o `False`).

Los operadores de **comparación** que se pueden utilizar son:
- Igualdad `==`
- Desigualdad `!=`
- Menor que `<`
- Menor o igual que `<=`
- Mayor que `>`
- Mayor o igual que `>=`

```python
edad_juan = 30
edad_maria = 30
edad_pedro = 35
```

```python
edad_juan == edad_maria
```

```python
edad_juan == edad_pedro
```

```python
edad_juan != edad_pedro
```

```python
edad_juan < edad_maria
```

```python
edad_juan <= edad_maria
```

```python
edad_juan >= edad_pedro
```

Estos operadores de comparación se pueden combinar entre sí para formar expresiones más complejas. Tres de los operadores más comunmente utilizados son:

* **not**: niega el valor booleano
* **and**: conjunción (todo debe ser cierto para que el resultado sea cierto)
* **or**: disyunción (al menos un elemento debe ser cierto para que el resultado sea cierto)

Por ejemplo, disponemos de una edad y queremos saber si es mayor de edad (18 años):

```python
edad = 55
genero = "mujer"
```

```python
mayor_de_edad = edad >= 18
```

```python
mayor_de_edad
```

```python
type(mayor_de_edad)
```

Ahora queremos saber si, además de ser mayor de edad, es mujer o no:

```python
mayor_de_edad_y_mujer = mayor_de_edad and genero == "mujer"
```

```python
mayor_de_edad_y_mujer
```

```python
type(mayor_de_edad_y_mujer)
```

O si es menor de edad o es hombre (si una de las dos condiciones se cumple):

```python
menor_de_edad_u_hombre = not mayor_de_edad or genero == "hombre"
```

```python
menor_de_edad_u_hombre
```

En este último ejemplo, la jerarquía o precedencia de operadores es: primero las negaciones, luego las operaciones de conjunción y por último las operaciones de disyunción. No obstante, podemos usar paréntesissi nos ayuda a leerlo más fácilmente:

```python
(not mayor_de_edad) or (genero == "hombre")
```

Otro caso de uso muy demandado es comprobar si un valor está dentro de un rango:

```python
edad_minima = 20
edad_maxima = 60
edad = 18
```

```python
(edad >= edad_minima) and (edad <= edad_maxima)
```

Como es una operación tan habitual, existe la forma de representarla de una forma más legible y compacta:

```python
edad_minima <= edad <= edad_maxima
```
