# Clases y Objetos

Una clase es una definición que consiste en una colección de **variables** y **métodos** (funciones). Un objeto es una **instanciación** de esa clase, es decir, un objeto que contiene esas variables y funciones. La clase es la receta (solo definición), mientras que el objeto es algo manipulable (podemos llamar a sus funciones, sus atributos, etc.)

Dos objetos creados a partir de la misma clase son completamente **independientes**. Para referirse al propio objeto dentro de la definición de las clases, se usa (_por convenio_) la variable `self`.

La definición de clases se hace con la palabra reservada `class`. Existe una función que será ejecutada siempre que un objeto se cree a partir de la clase, llamada `__init__`. En una clase, toda función tiene como primer parámetro un argumento llamado `self` (salvo si se usa el decorador @staticmethod: más información [aquí](https://docs.python.org/3/library/functions.html#staticmethod)). Como mencionamos antes, este parámetro permite referenciar al propio objeto dentro de la definición de la clase. Para ver información más detallada sobre las clases, y propiedades como la _herencia_, consultar [aquí](https://docs.python.org/3.6/tutorial/classes.html).

Para crear un objeto a partir de una clase, se usa la misma sintaxis que en la llamada a funciones. De hecho, es como llamar a la función `__init__`, omitiendo el primer parámetro `self` y el nombre de la función (`__init__`):

```python
class MiClase:
    def __init__(self, x):
        self.x = x
        print("¡Objeto creado!")

mi_objeto = MiClase(5)
```

```python
# Podemos definir una clase con una función
class MiClase:
    def saludar(self):
        print("¡Hola!")
mi_objeto = MiClase()
mi_objeto.saludar()  # Invocamos la función saludar del objeto
```

<font size=4> <i class="fa fa-book" aria-hidden="true" style="color:#00586D"></i> </font> __Importante!__ En el ejemplo a continuación es necesario realizar la asignación `self.nombre = nombre` para que el nombre se guarde en el objeto. Si no, sería una variable declarada en el ámbito de la función que se perdería en cuanto finalizase la ejecución de la misma.

```python
# Podemos definir una clase con una función y una variable, que además se inicializa en el __init__
class MiClase:
    def __init__(self, nombre):
        self.nombre = nombre # Es necesario 
    def saludar(self):
        print("¡Hola {}!".format(self.nombre))
mi_objeto = MiClase("Javi")
mi_objeto.saludar()  # Invocamos la función saludar del objeto

# Podemos acceder a los atributos (variables almacenadas en self)
print(mi_objeto.nombre)
```

Terminología orientada a objetos
===
Las clases son parte de un paradigma de programación llamado **programación orientada a objetos**. La programación orientada a objetos, o OOP para abreviar, se enfoca en construir bloques de código reutilizables llamados clases. Cuando desea utilizar una clase en uno de sus programas, crea un **objeto** de esa clase, que es de donde proviene la frase "orientado a objetos". Python en sí no está vinculado a la programación orientada a objetos, pero usará objetos en la mayoría o en todos sus proyectos de Python. Para comprender las clases, debe comprender parte del lenguaje que se usa en OOP.

Terminología general
---
Una **clase** es un cuerpo de código que define los **atributos** y **comportamientos** necesarios para modelar con precisión algo que necesita para su programa. Puede modelar algo del mundo real, como un cohete espacial o una cuerda de guitarra, o puede modelar algo de un mundo virtual, como un cohete en un juego, o un conjunto de leyes físicas para un motor de juego.

Un **atributo** es un dato. En el código, un atributo es solo una variable que forma parte de una clase.

Un **comportamiento** es una acción que se define dentro de una clase. Estos se componen de **métodos**, que son solo funciones que se definen para la clase.

Un **objeto** es una instancia particular de una clase. Un objeto tiene un determinado conjunto de valores para todos los atributos (variables) de la clase. Puede tener tantos objetos como desee para cualquier clase.

Aceptando parámetros para el método \_\_init\_\_()
---
El método \_\_init\_\_() se ejecuta automáticamente una vez cuando crea un nuevo objeto a partir de una clase. El método \_\_init\_\_() para la clase Rocket hasta ahora es bastante simple:

```python
class Rocket():
    # La clase Rocket simula un cohete
    def __init__(self):
        # Cada cohete tiene una posición (x,y)
        self.x = 0
        self.y = 0
        
    def move_up(self):
        # Incremento de la posición y
        self.y += 1
```

Todo lo que hace el método \_\_init\_\_() hasta ahora es establecer los valores xey para el cohete en 0. Podemos agregar fácilmente un par de argumentos de palabras clave para que los nuevos cohetes puedan inicializarse en cualquier posición:

```python
class Rocket():
    # La clase Rocket simula un cohete
    def __init__(self, x=0, y=0):
        # Cada cohete tiene una posición (x,y)
        self.x = x
        self.y = y
        
    def move_up(self):
        # Incremento de la posición y
        self.y += 1
```

Ahora, cuando crea un nuevo objeto Rocket, tiene la opción de pasar valores iniciales arbitrarios para x e y:

```python
# creamos una serie de posiciones de inicio diferentes
rockets = []
rockets.append(Rocket())
rockets.append(Rocket(0,10))
rockets.append(Rocket(100,0))

# Mostrar cada uno de los cohetes
for index, rocket in enumerate(rockets):
    print("Rocket %d is at (%d, %d)." % (index, rocket.x, rocket.y))
```

Aceptar parámetros en un método
---
El método \_\_init\_\_ es solo un método especial que tiene un propósito particular, que es ayudar a crear nuevos objetos a partir de una clase. Cualquier método de una clase puede aceptar parámetros de cualquier tipo. Con esto en mente, el método move_up() se puede hacer mucho más flexible. Al aceptar argumentos de palabras clave, el método move_up() se puede reescribir como un método move_rocket() más general. Este nuevo método permitirá que el cohete se mueva cualquier cantidad, en cualquier dirección:

```python
class Rocket():
    # La clase Rocket simula un cohete
    def __init__(self, x=0, y=0):
        # Cada cohete tiene una posición (x,y)
        self.x = x
        self.y = y
        
    def move_rocket(self, x_increment=0, y_increment=1):
        # Movermos el cohete dependiendo de los parametros
        # por defecto moveremos el cohete una unidad hacia arriba
        self.x += x_increment
        self.y += y_increment
    
    # método interno para mostrar la información de esta clase con tostring()
    def __repr__(self):
        return f"Rocket en posición x: {self.x} y: {self.y}"
```

```python
# Creamos una lista de cohetes
rockets = []
for x in range(0,3):
    rockets.append(Rocket())

# Movemos cada cohete una cantidad diferente
rockets[0].move_rocket()
rockets[1].move_rocket(10,10)
rockets[2].move_rocket(-10,0)
          
# Mostramos donde está el cohete
for index, rocket in enumerate(rockets):
    print("Rocket %d is at (%d, %d)." % (index, rocket.x, rocket.y))
```

## Uso del método **\_\_repr\_\_()**

```python
class Rocket():
    # La clase Rocket simula un cohete
    def __init__(self, x=0, y=0):
        # Cada cohete tiene una posición (x,y)
        self.x = x
        self.y = y
        
    def move_rocket(self, x_increment=0, y_increment=1):
        # Movermos el cohete dependiendo de los parametros
        # por defecto moveremos el cohete una unidad hacia arriba
        self.x += x_increment
        self.y += y_increment
    
    # método interno para mostrar la información de esta clase con tostring()
    def __repr__(self):
        return f"Rocket en posición x: {self.x} y: {self.y}"
```

```python
# Creamos una lista de cohetes
rockets = []
for x in range(0,3):
    rockets.append(Rocket())
    
# Movemos cada cohete una cantidad diferente
rockets[0].move_rocket()
rockets[1].move_rocket(10,10)
rockets[2].move_rocket(-10,0)

for rocket in rockets:
    print(rocket)
```
